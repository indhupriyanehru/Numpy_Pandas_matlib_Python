```python
# STUDENT PERFORMANCE ANALYSIS SYSTEM
# Using NumPy, Pandas and Matplotlib


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------
# 1. STUDENT DATA
# ------------------------------------------

students = ["Arun", "Priya", "Rahul", "Sneha", "Kiran","Divya", "Ravi", "Anjali", "Vijay", "Meena"]

subjects = ["English", "Math", "Science", "Social", "Spanish"
]

marks = [
    [85, 90, 78, 88, 92],
    [92, 95, 89, 91, 87],
    [76, 80, 85, 79, 82],
    [88, 86, 92, 90, 95],
    [70, 75, 72, 78, 80],
    [95, 98, 96, 94, 97],
    [82, 84, 80, 85, 88],
    [90, 92, 94, 89, 91],
    [65, 70, 68, 72, 75],
    [87, 89, 85, 90, 88]
]

# ------------------------------------------
# 2. NUMPY OPERATIONS
# ------------------------------------------

print("\n========== NUMPY OPERATIONS ==========")

arr = np.array(marks)

# Display array
print("\nStudent Marks Array:\n", arr)

# Array shape and dimensions
print("\nShape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Total Elements:", arr.size)

# Indexing
print("\nMarks of first student:", arr[0])
print("English marks of all students:", arr[:, 0])
print("Math marks of all students:", arr[:, 1])

# Slicing
print("\nFirst 3 students:\n", arr[:3])
print("First 3 subjects:\n", arr[:, :3])

# Maximum and minimum marks
print("\nMaximum marks in each subject:",
      np.max(arr, axis=0))

print("Minimum marks in each subject:",
      np.min(arr, axis=0))

# Mean and median
print("\nMean of each subject:",
      np.mean(arr, axis=0))

print("Median of each subject:",
      np.median(arr, axis=0))

# Average marks of each student
student_average = np.mean(arr, axis=1)

print("\nStudent Averages:", student_average)

# Overall class average
print("\nOverall Class Average:",
      np.mean(arr))

# Standard deviation and variance
print("\nStandard Deviation:",
      np.std(arr, axis=0))

print("Variance:",
      np.var(arr, axis=0))

# Sorting
print("\nSorted English Marks:",
      np.sort(arr[:, 0]))

# Argmax and argmin
for i, subject in enumerate(subjects):
    highest_index = np.argmax(arr[:, i])
    lowest_index = np.argmin(arr[:, i])

    print(f"\n{subject}")
    print("Highest scorer:", students[highest_index])
    print("Highest marks:", arr[highest_index, i])
    print("Lowest scorer:", students[lowest_index])
    print("Lowest marks:", arr[lowest_index, i])

# ------------------------------------------
# 3. PANDAS OPERATIONS
# ------------------------------------------

print("\n========== PANDAS OPERATIONS ==========")

df = pd.DataFrame(
    arr,
    index=students,
    columns=subjects
)

# Add total and average columns
df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1)

print("\nComplete Student DataFrame:\n")
print(df)

# Display first and last records
print("\nFirst 5 Students:\n", df.head())
print("\nLast 5 Students:\n", df.tail())

# DataFrame information
print("\nDataFrame Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:\n")
print(df.describe())

# Access a particular student
print("\nArun's Marks:\n", df.loc["Arun"])

# Access using integer position
print("\nFirst Student:\n", df.iloc[0])

# Access a particular subject
print("\nEnglish Marks:\n", df["English"])

# Filtering students
print("\nStudents scoring above 90 in Math:")
print(df[df["Math"] > 90])

# Students with average above 85
print("\nStudents with average above 85:")
print(df[df["Average"] > 85])

# Sorting students by total
print("\nStudents sorted by total marks:")
print(df.sort_values(by="Total", ascending=False))

# Subject-wise statistics
print("\nSubject-wise Maximum:")
print(df[subjects].max())

print("\nSubject-wise Minimum:")
print(df[subjects].min())

print("\nSubject-wise Mean:")
print(df[subjects].mean())

print("\nSubject-wise Median:")
print(df[subjects].median())

# Rank students
df["Rank"] = df["Total"].rank(
    ascending=False,
    method="min"
)

print("\nStudent Ranking:\n")
print(df.sort_values("Rank"))

# Grouping students into grades
def calculate_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average"].apply(calculate_grade)

print("\nStudents with Grades:\n")
print(df[["Total", "Average", "Grade"]])

# Group by grade
print("\nGrade-wise Student Count:")
print(df.groupby("Grade").size())

# Save data into CSV
df.to_csv("student_marks.csv")

print("\nCSV file created successfully!")

# ------------------------------------------
# 4. MATPLOTLIB VISUALIZATION
# ------------------------------------------

print("\n========== MATPLOTLIB GRAPHS ==========")

# 1. Bar chart - Average marks of students
plt.figure(figsize=(10, 5))

plt.bar(students, student_average, color="skyblue")

plt.title("Average Marks of Students")
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2. Line graph - Subject-wise average
subject_average = np.mean(arr, axis=0)

plt.figure(figsize=(8, 5))

plt.plot(subjects, subject_average,
         marker="o", color="green")

plt.title("Subject-wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.grid(True)
plt.show()

# 3. Pie chart - Subject-wise average
plt.figure(figsize=(7, 7))

plt.pie(
    subject_average,
    labels=subjects,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Subject-wise Average Distribution")
plt.show()

# 4. Histogram - Distribution of marks
plt.figure(figsize=(8, 5))

plt.hist(arr.flatten(), bins=10,
         color="orange", edgecolor="black")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

# 5. Scatter plot - English vs Math
plt.figure(figsize=(8, 5))

plt.scatter(arr[:, 0], arr[:, 1],
            color="red")

plt.title("English vs Mathematics Marks")
plt.xlabel("English")
plt.ylabel("Mathematics")
plt.grid(True)
plt.show()

# 6. Subject-wise comparison
df[subjects].mean().plot(
    kind="bar",
    color="purple",
    figsize=(8, 5)
)

plt.title("Comparison of Average Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.tight_layout()
plt.show()

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
